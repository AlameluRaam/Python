```python
marks = 91
if marks > 90:
    print ("Greater than 90")
    if marks >95:  ## nested IF statement.
        print("marks is greater than 95")
    else:
        print ("marks is less than 95")
elif marks < 90 and marks > 50:
    print("marks is in between 50 to 90")
elif marks < 50 and marks > 30:
    print("marks is in between 30 to 50")
else:
    print("Marks is less than 30")
```
