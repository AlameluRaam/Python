```python
L1 = [1,2,3,4,5]
L2 = [101, 102, 103, 104, 105]

Z = zip(L1, L2)
for x in Z:
    print(x, end= ",")
```

```python
L1 = [1,2,3,4,5]
L2 = [101, 102, 103, 104, 105]

Z = zip(L1, L2)
for (x,y) in Z:
    sum = x+y
    print(sum, end= ",")
```

```python
L1 = [1,2,3,4,5]
L2 = [101, 102, 103, 104, 105]
L3 = [6,7,8,9,10]
L4 = [105, 106, 107, 108, 109]

Z = zip(L1, L2, L3, L4)
for (a,b,c,d) in Z:
    print(a,b,c,d, sep=",", end= "\n")
```

```python
L1 = ['a','b','c','d','e']
L2 = [101, 102, 103, 104, 105]

D = dict(zip(L1, L2))
for x in D.items():
    print(x, end= ",")
print("\n")
print(D)
```
