```python
def fnMaker():
    def displayFn():
        displayFn.count = displayFn.count + 1
        print("Inside nested function ", displayFn.count )
    displayFn.count = 100
    print("Outside nested function ", displayFn.count )
    return displayFn

A = fnMaker()
print("Outsise all function ", A.count)
```

## Generator Function

```python
def generatorFn():
    L = [10,20,30,40]
    for x in L:
        yield x

X = generatorFn()
print(next(X))
```

## Extended Generator Function

```python
def generatorFn():
    yield "First return" 
    result1 = yield "Second return" 
    print("Inside generator function 1 ", result1)
    result2 = yield "Third return" ##<<--
    print("Inside generator function 2", result2)

X = generatorFn()
print(next(X))
```
