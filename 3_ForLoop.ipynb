```python
D = {'a':1, 'b':2, 'c':3}
for x in D.keys():
    print(x, end=",")
print("\n")
for x in D.values():
    print(x, end=",")
print("\n")
for x in D.items():
    print(x, end=",")
print("\n")
for (x,y) in D.items():
    print(x,y, sep=" ", end=",")
print("\n")
```
